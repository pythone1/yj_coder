# 哨兵二号

```python
# 影像查询中心点
center = [119.46876, 34.54992]
point_geom = ee.Geometry.Point(center)

# 时间范围
date_range = ['2023-01-01','2025-03-10']

# 最大云量【默认100，取所有影像，下载处理后再根据实际情况筛选】
cld_pct = 10
```

```python
# 查找数据集(反射率)
imgcoll = ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED')\
    .filterBounds(point_geom)\
    .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE',cld_pct))\
    .filterDate(date_range[0],date_range[1])\
    .select(['B2','B3','B4','B8','B11','B12'])
    # .select(['B1','B2','B3','B4','B5','B6','B7','B8','B8A','B9','B11','B12'])
    
# 提取数据集中影像数量及产品ID信息
img_num = imgcoll.size().getInfo()
product_list = imgcoll.reduceColumns(ee.Reducer.toList(),['PRODUCT_ID']).get("list").getInfo()# 可查询的属性信息，通过取collection中的一景图像，查询其属性名获取，如latest_img.propertyNames().getInfo()
print("共找到",img_num,'张影像，产品编号如下：')
for i in range(img_num):
    print(i,':',product_list[i])
```

```python
# 确认影像编号
img_idx = 1

# 获取指定图像
img_list = imgcoll.toList(img_num)
latest_img = ee.Image(img_list.get(img_idx))

# 哨兵二号RGB显示样式
s2_rgb_vis_params = {
    'min':0,
    'max':5000,
    'bands':['B4','B3','B2']
}

#显示指定图像
gmap = geemap.Map(center=(center[1],center[0]),zoom=10)
gmap.addLayer(latest_img,s2_rgb_vis_params,'sentinel-2')
gmap
```
